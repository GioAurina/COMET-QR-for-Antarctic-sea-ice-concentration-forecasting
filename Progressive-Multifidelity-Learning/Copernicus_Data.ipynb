```python
import copernicusmarine
import xarray as xr
import os
import numpy as np

# --- CONFIGURAZIONE ---
dataset_id = "cmems_mod_glo_phy_my_0.083deg_P1D-m"
output_folder = "./dati_antartide_SST_raw"  # Nuova cartella per la temperatura
years = range(1993, 2024)

if not os.path.exists(output_folder):
    os.makedirs(output_folder)

print("--- DOWNLOAD SEA SURFACE TEMPERATURE (SST) ---")

for year in years:
    raw_filename = f"temp_sst_raw_{year}.nc"
    final_filename = f"antartide_sst_{year}.nc"
    full_path_final = os.path.join(output_folder, final_filename)
    
    if os.path.exists(full_path_final):
        print(f"Anno {year} già fatto. Salto.")
        continue

    print(f"\nElaborazione {year}...", end="")
    
    try:
        # 1. DOWNLOAD
        copernicusmarine.subset(
            dataset_id=dataset_id,
            dataset_version="202311",
            username="",
            password="", 
            
            # AREA (Stessa dell'altra volta)
            minimum_latitude=-90,
            maximum_latitude=-50,
            minimum_longitude=-180,
            maximum_longitude=180,
            
            # TEMPO
            start_datetime=f"{year}-01-01 00:00:00",
            end_datetime=f"{year}-12-31 23:59:59",
            
            # --- MODIFICHE CRUCIALI PER SST ---
            variables=["thetao"],       # 'thetao' è la temperatura potenziale
            minimum_depth=0,            # Inizio dalla superficie
            maximum_depth=1,            # Mi fermo subito (prendo solo il layer 0.49m)
            
            output_filename=raw_filename,
            force_download=True
        )

        # 2. COMPRESSIONE E PULIZIA
        with xr.open_dataset(raw_filename) as ds:
            # Rimuoviamo la coordinata depth che ora è inutile (è size=1)
            ds = ds.squeeze(drop=True) 
            
            encoding = {
                'thetao': {
                    'zlib': True, 
                    'complevel': 5, 
                    'dtype': 'float32', 
                    '_FillValue': np.nan 
                }
            }
            ds.to_netcdf(full_path_final, encoding=encoding)
        
        if os.path.exists(raw_filename):
            os.remove(raw_filename)
            
        print(" Fatto!")

    except Exception as e:
        print(f" ERRORE: {e}")
        if os.path.exists(raw_filename): os.remove(raw_filename)
```

```python
import xarray as xr
import numpy as np
import os
from pyproj import Transformer

# --- CONFIGURAZIONE ---
input_folder = "./dati_antartide_SST_raw"      # Input SST
output_folder = "./dati_antartide_SST_regrid"  # Output SST Regriddato
years = range(1993, 2024)

# GRIGLIA 432x432 (La tua specifica)
grid_size = 432      
pixel_res = 25000    
half_width = (grid_size * pixel_res) / 2 

x = np.linspace(-half_width, half_width, grid_size)
y = np.linspace(-half_width, half_width, grid_size)
target_x, target_y = np.meshgrid(x, y)

transformer = Transformer.from_crs("EPSG:3031", "EPSG:4326", always_xy=True)
target_lon, target_lat = transformer.transform(target_x, target_y)

tgt_lat_da = xr.DataArray(target_lat, dims=("y", "x"))
tgt_lon_da = xr.DataArray(target_lon, dims=("y", "x"))

if not os.path.exists(output_folder):
    os.makedirs(output_folder)

print("--- REGRIDDING SST (432x432) ---")

for year in years:
    input_file = os.path.join(input_folder, f"antartide_sst_{year}.nc")
    output_file = os.path.join(output_folder, f"antartide_sst_{year}_regrid.nc")
    
    if not os.path.exists(input_file): continue
    if os.path.exists(output_file): continue

    print(f"Elaborazione {year}...", end="")

    try:
        with xr.open_dataset(input_file) as ds:
            # Interpolazione
            regridded = ds.interp(
                latitude=tgt_lat_da, 
                longitude=tgt_lon_da, 
                method="linear"
            )
            
            # Gestione Terraferma
            # Mettiamo 0 dove c'era NaN (terra). 
            # L'acqua sarà negativa (es. -1.8) o positiva (estate).
            regridded = regridded.fillna(0)
            
            # Salvataggio
            encoding = {'thetao': {'zlib': True, 'dtype': 'float32'}}
            regridded.to_netcdf(output_file, encoding=encoding)
            
        print(" OK")

    except Exception as e:
        print(f" ERRORE: {e}")

print("Finito.")
```

```python
import dill
import xarray as xr
import numpy as np
import os

# --- CONFIGURAZIONE ---
input_folder = "./dati_antartide_SST_regrid" # Prende dalla cartella nuova
output_pickle = "./Antarctic_SST_1993_2023_432.pkl"
years = range(1993, 2024)

data_SST_list = []
years_loaded = []

print("--- CREAZIONE LISTA DATI (432x432) ---")

for year in years:
    filename = f"antartide_sst_{year}_regrid.nc"
    file_path = os.path.join(input_folder, filename)
    
    if os.path.exists(file_path):
        print(f"Caricamento {year}...", end="")
        
        with xr.open_dataset(file_path) as ds:
            year_data = ds['thetao'].values.astype(np.float32)
            
            # --- CONTROLLO CORRETTO ---
            spatial_shape = year_data.shape[1:] 
            
            # Ora controlliamo che sia 432x432
            if spatial_shape == (432, 432):
                data_SST_list.append(year_data)
                years_loaded.append(year)
                print(f" OK! ({year_data.shape[0]} giorni)")
            else:
                print(f" ❌ ERRORE: Shape {year_data.shape} errata!")
                
    else:
        print(f"Anno {year} mancante.")

print("\n--- SALVATAGGIO PICKLE ---")
try:
    with open(output_pickle, 'wb') as f:
        dill.dump((data_SST_list, years_loaded), f)
    print(f"✅ File salvato: {output_pickle}")
    
except Exception as e:
    print(f"❌ Errore salvataggio: {e}")
```
